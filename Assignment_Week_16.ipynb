```python
!pip install openai langchain tiktoken

import os
from langchain.chat_models import ChatOpenAI
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate

import getpass
os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter your OpenAI API Key: ")

prompt_template = PromptTemplate(
    input_variables=["recipient", "subject", "tone", "key_points"],
    template="""
    You are an assistant that writes professional and secure emails.

    Write an email to {recipient}.
    Subject: {subject}
    Tone: {tone}
    Key Points: {key_points}

    Make the email sound human-like, engaging, and well-structured.
    """
)

llm = ChatOpenAI(
    temperature=0.7,  # Adjusts creativity
    model_name="gpt-4",  # or gpt-3.5-turbo if GPT-4 not available
    max_tokens=500
)

email_chain = LLMChain(
    llm=llm,
    prompt=prompt_template
)

response = email_chain.run({
    "recipient": "John from Finance",
    "subject": "Quarterly Budget Review",
    "tone": "Formal and Friendly",
    "key_points": "Thank him for the report, highlight a budget discrepancy, request a follow-up meeting"
})

print(response)
```
