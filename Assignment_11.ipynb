```python
from chatterbot import ChatBot
from chatterbot.trainers import ChatterBotCorpusTrainer, ListTrainer

# Create a new chatbot instance
chatbot = ChatBot(
    "MediBot",
    storage_adapter="chatterbot.storage.SQLStorageAdapter",
    logic_adapters=["chatterbot.logic.BestMatch"],
    database_uri="sqlite:///database.sqlite3"
)

# Corpus training
corpus_trainer = ChatterBotCorpusTrainer(chatbot)
corpus_trainer.train("chatterbot.corpus.english")

# Custom hospital chatbot conversation
conversation = [
    "Hi",
    "Hello! I am MediBot, your hospital assistant. How can I help you today?",
    "I want to book an appointment",
    "Sure! Which department do you need? (Cardiology, Pediatrics, Orthopedics, etc.)",
    "Pediatrics",
    "We have Dr. Chen and Dr. Smith available. Who would you like to book with?",
    "Dr. Chen",
    "Available time slots for Dr. Chen are 10 AM, 11 AM, and 2 PM. Please choose one.",
    "11 AM",
    "Booking confirmed with Dr. Chen at 11 AM. Would you like a reminder?",
    "Yes",
    "Reminder set. Anything else I can help you with?",
    "What are your visiting hours?",
    "We are open Monday to Saturday, 8 AM to 6 PM.",
    "Emergency contact?",
    "For emergencies, please call +1-800-123-4567.",
    "Thank you",
    "You're welcome! Stay healthy."
]

# Custom training
trainer = ListTrainer(chatbot)
trainer.train(conversation)
```

```python
def chat_with_bot():
    print("MediBot: Hello! Type 'exit' to end the chat.\n")
    while True:
        user_input = input("You: ")
        if user_input.lower() in ['exit', 'quit', 'bye']:
            print("MediBot: Goodbye! Stay healthy.")
            break
        response = chatbot.get_response(user_input)
        print("MediBot:", response)

chat_with_bot()
```
